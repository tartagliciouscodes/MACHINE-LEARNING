DATASET

```python
import pandas as pd

data = {
    'Variable': ['Climate', 'Housing', 'Health', 'Crime', 'Transportation',
                 'Education', 'Arts', 'Recreation', 'Economy'],
    'PC1': [0.190, 0.544, 0.782, 0.365, 0.585, 0.394, 0.985, 0.520, 0.142],
    'PC2': [0.017, 0.020, -0.605, 0.294, 0.085, -0.273, 0.126, 0.402, 0.150],
    'PC3': [0.207, 0.204, 0.144, 0.585, 0.234, 0.027, -0.111, 0.519, 0.239]
}

df = pd.DataFrame(data)
df
```

FIND VARIABLE WITH HIGHEST LOADING

```python
for pc in ['PC1', 'PC2', 'PC3']:
    index = df[pc].abs().idxmax()
    print(pc, "->", df.loc[index, 'Variable'],
          "(", df.loc[index, pc], ")")
```

DISPLAY AS A TABLE

```python
result = []

for pc in ['PC1', 'PC2', 'PC3']:
    index = df[pc].abs().idxmax()
    result.append([pc, df.loc[index, 'Variable'], df.loc[index, pc]])

result_df = pd.DataFrame(
    result,
    columns=['Principal Component', 'Variable', 'Loading']
)

result_df
```

DISPLAY PCA LOADING GRAPH

```python
import matplotlib.pyplot as plt

df.plot(
    x='Variable',
    y=['PC1', 'PC2', 'PC3'],
    kind='bar',
    figsize=(12, 6)
)

plt.xlabel('Variable')
plt.ylabel('Loading')
plt.title('Principal Component Loadings')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
