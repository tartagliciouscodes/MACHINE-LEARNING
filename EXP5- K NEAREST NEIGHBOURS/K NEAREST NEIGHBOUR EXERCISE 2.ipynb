DATASET

```python
import pandas as pd
import numpy as np

data = {
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
}

df = pd.DataFrame(data)
df
```

NEW VALUE

```python
new_brightness = 45
new_saturation = 40
```

CALCULATE EUCLIDEAN DISTANCE

```python
df['Distance'] = np.sqrt(
    (df['Brightness'] - new_brightness)**2 +
    (df['Saturation'] - new_saturation)**2
)

df
```

DISPLAY DISTANCE

```python
print("Distances from new entry:")
print(df[['Brightness', 'Saturation', 'Class', 'Distance']])
```

SORT DISTANCE

```python
df_sorted = df.sort_values('Distance')

df_sorted[['Brightness', 'Saturation', 'Class', 'Distance']]
```

DISPLAY KNN CLASSIFICATION

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

for class_name in df['Class'].unique():
    subset = df[df['Class'] == class_name]
    plt.scatter(
        subset['Brightness'],
        subset['Saturation'],
        label=class_name,
        s=100
    )

# New entry
plt.scatter(
    new_brightness,
    new_saturation,
    marker='X',
    s=200,
    label='New Entry'
)

plt.xlabel('Brightness')
plt.ylabel('Saturation')
plt.title('Brightness vs Saturation')
plt.legend()
plt.grid(True)
plt.show()
```
