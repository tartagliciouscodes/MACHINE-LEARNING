DECISION TREE CLASSIFIER EXERCISE-2

IMPORTING LIBRARIES

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

DATASET

```python
data = {
    "Weather": ["Sunny", "Sunny", "Rainy", "Rainy", "Sunny", "Rainy"],
    "Decision": ["Go", "Go", "Stay", "Stay", "Go", "Stay"]
}

df = pd.DataFrame(data)

df
```

PREPARE DATA

```python
X = pd.get_dummies(df["Weather"])
y = df["Decision"]

X
```

TRAIN DECISION TREE

```python
model = DecisionTreeClassifier(criterion="entropy")
model.fit(X, y)

print("Accuracy:", model.score(X, y))
```

PREDICT

```python
weather = input("Enter weather (Sunny/Rainy): ")

new_data = pd.DataFrame({"Weather": [weather]})
new_data = pd.get_dummies(new_data)
new_data = new_data.reindex(columns=X.columns, fill_value=0)

prediction = model.predict(new_data)

print("Decision:", prediction[0])
```

DISPLAY DECISION TREE

```python
plt.figure(figsize=(7, 5))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=model.classes_,
    filled=True,
    rounded=True,
    fontsize=8
)

plt.tight_layout()
plt.show()
```
